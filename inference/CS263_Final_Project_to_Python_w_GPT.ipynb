# Implementing CodeBleu to Evaluation Similarities

````python
# CodeBLEU =
# α * BLEU + β * weighted n-gram + γ * AST match + δ * data-flow match

# Each component captures:

# BLEU: token-level n-gram match

# Weighted n-gram: gives higher weight to important keywords

# AST match: structural similarity based on the Abstract Syntax Tree

# Data-flow match: checks semantic consistency in variable logic
````

````python
import sacrebleu
from tree_sitter import Parser
from tree_sitter_languages import get_language
import zss
from tqdm import tqdm

# Define keywords for weighted n-gram match
KEYWORDS = {'def', 'class', 'return', 'if', 'else', 'for', 'while', 'try', 'except', 'with', 'import', 'from', 'as'}

def tokenize_code(code):
    if not isinstance(code, str):
        return []
    return code.strip().replace("\n", " ").split()

def weighted_ngram_score(hyp, ref):
    hyp_tokens = tokenize_code(hyp)
    ref_tokens = set(tokenize_code(ref))
    match = sum(5 if tok in KEYWORDS else 1 for tok in hyp_tokens if tok in ref_tokens)
    total = sum(5 if tok in KEYWORDS else 1 for tok in hyp_tokens)
    return match / total if total else 0

class ASTNode:
    def __init__(self, node):
        self.label = node.type
        self.children = [ASTNode(child) for child in node.children]

    def get_children(self):
        return self.children

    def get_label(self):
        return self.label

def tree_edit_sim_score(hyp_code, ref_code, parser, tree_cache):
    try:
        if hyp_code not in tree_cache:
            tree_cache[hyp_code] = parser.parse(bytes(hyp_code, "utf8")).root_node
        if ref_code not in tree_cache:
            tree_cache[ref_code] = parser.parse(bytes(ref_code, "utf8")).root_node

        hyp_tree = tree_cache[hyp_code]
        ref_tree = tree_cache[ref_code]

        hyp_ast = ASTNode(hyp_tree)
        ref_ast = ASTNode(ref_tree)

        dist = zss.simple_distance(
            hyp_ast, ref_ast,
            get_children=lambda node: node.get_children(),
            get_label=lambda node: node.get_label()
        )

        max_size = max(hyp_tree.end_byte, ref_tree.end_byte)
        return 1 - dist / max_size if max_size else 0
    except Exception:
        return 0

def run_codebleu(refs, hyps, lang="python"):
    refs = [str(r) if r is not None else "" for r in refs]
    hyps = [str(h) if h is not None else "" for h in hyps]

    # BLEU
    if all(r == "" for r in refs) or all(h == "" for h in hyps):
        bleu = 0.0
    else:
        bleu = sacrebleu.corpus_bleu(hyps, [refs]).score

    # Weighted n-gram match
    weighted_scores = [weighted_ngram_score(h, r) for h, r in zip(hyps, refs)]
    weighted_bleu = sum(weighted_scores) / len(weighted_scores) * 100

    # AST match
    parser = Parser()
    parser.set_language(get_language(lang))
    tree_cache = {}

    ast_scores = []
    for h, r in tqdm(zip(hyps, refs), total=len(hyps), desc="Computing AST scores"):
        ast_scores.append(tree_edit_sim_score(h, r, parser, tree_cache))
    ast_score = sum(ast_scores) / len(ast_scores) * 100

    # Print results
    print(f"BLEU: {round(bleu, 2)}")
    print(f"Weighted n-gram match: {round(weighted_bleu, 2)}")
    print(f"AST match (tree edit similarity): {round(ast_score, 2)}")

    return {
        "bleu": bleu,
        "weighted": weighted_bleu,
        "ast": ast_score
    }
````

````python
# Recreate ref/hyp lists from your DataFrame
refs = df_filtered["python_code"].apply(lambda x: x.strip().replace("\n", " ")).tolist()
hyps = df_filtered["parsed_python_from_javascript"].apply(lambda x: x.strip().replace("\n", " ")).tolist()

# Evaluate
scores = run_codebleu(refs, hyps)
````

````python
import os
import json
import pandas as pd
import ast
import math
import copy
import re
import builtins
import multiprocessing
import textwrap

# Helper class definitions and typing imports
helper_classes = """
from typing import List, Optional, Dict, Tuple, Set

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __eq__(self, other):
        a, b = self, other
        while a and b:
            if a.val != b.val:
                return False
            a, b = a.next, b.next
        return a is None and b is None

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
"""

def run_with_timeout(func, args=(), kwargs=None, timeout=2):
    if kwargs is None:
        kwargs = {}
    def wrapper(queue):
        try:
            result = func(*args, **kwargs)
            queue.put((True, result))
        except Exception as e:
            queue.put((False, e))

    queue = multiprocessing.Queue()
    p = multiprocessing.Process(target=wrapper, args=(queue,))
    p.start()
    p.join(timeout)

    if p.is_alive():
        p.terminate()
        p.join()
        return False, TimeoutError("Execution timed out")
    else:
        success, result = queue.get()
        return success, result

def flatten_listnodes(node):
    result = []
    while node:
        result.append(node.val)
        node = node.next
    return result

def list_to_tree(data):
    if not data or data[0] is None:
        return None
    root = TreeNode(data[0])
    queue = [root]
    i = 1
    while queue and i < len(data):
        current = queue.pop(0)
        if i < len(data) and data[i] is not None:
            current.left = TreeNode(data[i])
            queue.append(current.left)
        i += 1
        if i < len(data) and data[i] is not None:
            current.right = TreeNode(data[i])
            queue.append(current.right)
        i += 1
    return root

def list_to_listnode(lst, pos=-1):
    dummy = ListNode(0)
    current = dummy
    nodes = []
    for val in lst:
        node = ListNode(val)
        nodes.append(node)
        current.next = node
        current = current.next
    if pos != -1 and pos < len(nodes):
        current.next = nodes[pos]
    return dummy.next

def normalize_output(x):
    if isinstance(x, str):
        return x.strip().lower()
    if isinstance(x, float):
        return round(x, 6)
    if hasattr(x, 'val') and hasattr(x, 'next'):
        return flatten_listnodes(x)
    if isinstance(x, dict):
        return {k: normalize_output(v) for k, v in x.items()}
    return x

def normalize_list_of_lists(lst):
    return sorted([sorted(normalize_output(i) if isinstance(i, list) else i) for i in lst])

def smart_equal(a, b):
    a, b = normalize_output(a), normalize_output(b)
    if isinstance(a, str) and a.isdigit():
        a = int(a)
    if isinstance(b, str) and b.isdigit():
        b = int(b)
    if isinstance(a, float) and isinstance(b, float):
        return math.isclose(a, b, abs_tol=1e-6)
    if isinstance(a, bool) or isinstance(b, bool):
        return str(a).lower() == str(b).lower()
    if isinstance(a, list) and isinstance(b, list):
        try:
            if all(isinstance(x, list) for x in a + b):
                return normalize_list_of_lists(a) == normalize_list_of_lists(b)
            return sorted(normalize_output(x) for x in a) == sorted(normalize_output(x) for x in b)
        except Exception:
            return a == b
    return a == b

def unwrap_solution_class(code):
    if "class Solution" not in code:
        return code.strip()
    lines = code.splitlines()
    unwrapped = []
    inside_class = False
    for line in lines:
        if line.strip().startswith("class Solution"):
            inside_class = True
            continue
        if inside_class:
            if line.startswith("    "):
                unwrapped.append(line[4:])
            elif line.strip() == "":
                unwrapped.append("")
            else:
                break
    return "\n".join(unwrapped).strip()

def get_function_name_from_code(code):
    try:
        tree = ast.parse(code)
        for node in tree.body:
            if isinstance(node, ast.FunctionDef):
                return node.name
            elif isinstance(node, ast.ClassDef) and node.name == "Solution":
                for sub_node in node.body:
                    if isinstance(sub_node, ast.FunctionDef):
                        return sub_node.name
    except Exception:
        pass
    return None

def camel_to_snake(name):
    return re.sub(r'(?<!^)(?=[A-Z])', '_', name).lower()

def convert_arg_names(inputs, func):
    new_inputs = {}
    func_args = func.__code__.co_varnames[:func.__code__.co_argcount]
    manual_mapping = {"board": "piles", "nums": "piles"}
    for k, v in inputs.items():
        if k in func_args:
            new_inputs[k] = v
        else:
            k_snake = camel_to_snake(k)
            if k_snake in func_args:
                new_inputs[k_snake] = v
            elif k in manual_mapping and manual_mapping[k] in func_args:
                new_inputs[manual_mapping[k]] = v
    return new_inputs

def clean_strings(obj):
    if isinstance(obj, str):
        return obj.strip()
    elif isinstance(obj, list):
        return [clean_strings(x) for x in obj]
    elif isinstance(obj, dict):
        return {k: clean_strings(v) for k, v in obj.items()}
    return obj

def evaluate_function_robust(py_code: str, func_name: str, test_cases: list, is_method=False):
    global_env = {
        "__builtins__": builtins.__dict__,
        "List": list,
        "Optional": type(None),
        "Dict": dict,
        "Tuple": tuple,
        "Set": set
    }

    mismatches = []

    try:
        py_code = textwrap.dedent(py_code)
        full_code = helper_classes + "\n" + py_code
        exec(full_code, global_env)

        if is_method:
            solution_instance = global_env.get("Solution", lambda: None)()
            func = getattr(solution_instance, func_name, None)
        else:
            func = global_env.get(func_name)

        if not callable(func):
            return 0, len(test_cases), f"Function {func_name} not found", []
    except SyntaxError as se:
        return 0, len(test_cases), f"SyntaxError: {se}", []
    except Exception as e:
        return 0, len(test_cases), f"Exec error: {e}", []

    passed = 0
    for case in test_cases:
        try:
            inputs = clean_strings(copy.deepcopy(case["input"]))
            pos = inputs.get("pos", -1)

            for k, v in inputs.items():
                if isinstance(v, list) and k.lower() == "head":
                    inputs[k] = list_to_listnode(v, pos)
                elif isinstance(v, list) and k.lower() == "root":
                    inputs[k] = list_to_tree(v)
                elif isinstance(v, str):
                    v_clean = v.strip()
                    try:
                        if re.fullmatch(r"[01]+", v_clean):
                            inputs[k] = int(v_clean, 2)
                        elif v_clean.isdigit() and k.lower() not in ["a", "b"]:
                            inputs[k] = int(v_clean)
                        else:
                            inputs[k] = v_clean
                    except:
                        inputs[k] = v_clean

            args = convert_arg_names(inputs, func)
            success, result_or_error = run_with_timeout(func, args=(), kwargs=args, timeout=2)

            if not success:
                raise result_or_error

            actual = result_or_error
            if actual is None:
                for name in ["nums", "matrix"]:
                    if name in args:
                        actual = args[name]
                        break

            expected = clean_strings(case["output"])

            if isinstance(expected, str) and ',' in expected and any(x in expected for x in ['nums', 'matrix']):
                expected_num = int(expected.split(',')[0])
                expected_list = re.findall(r'\[(.*?)\]', expected)
                expected_list = [int(x) for x in expected_list[0].split(',') if x.strip().isdigit()] if expected_list else []
                result_ok = (actual == expected_num)
                actual_list = args.get("nums") or args.get("matrix")
                list_ok = (actual_list[:len(expected_list)] == expected_list)
                if result_ok and list_ok:
                    passed += 1
                    continue

            if smart_equal(actual, expected):
                passed += 1
            else:
                mismatches.append({
                    "input": case["input"],
                    "expected": case["output"],
                    "actual": actual
                })
        except Exception as e:
            mismatches.append({
                "input": case["input"],
                "expected": case["output"],
                "actual": f"Error: {e}"
            })

    return passed, len(test_cases), None, mismatches
````

````python
def evaluate_function_robust(py_code: str, func_name: str, test_cases: list, is_method=False):
    global_env = {"__builtins__": builtins.__dict__}
    mismatches = []

    try:
        # Inject helper classes into the environment
        exec(helper_classes, global_env)
        exec(py_code, global_env)

        if is_method:
            solution_instance = global_env.get("Solution", lambda: None)()
            func = getattr(solution_instance, func_name, None)
        else:
            func = global_env.get(func_name)

        if not callable(func):
            return 0, len(test_cases), f"Function {func_name} not found", []
    except Exception as e:
        return 0, len(test_cases), f"Exec error: {e}", []

    passed = 0
    for case in test_cases:
        try:
            inputs = clean_strings(copy.deepcopy(case["input"]))
            pos = inputs.get("pos", -1)

            # Convert inputs based on type
            for k, v in inputs.items():
                if isinstance(v, list):
                    if k.lower() == "head":
                        inputs[k] = list_to_listnode(v, pos)
                    elif k.lower() == "root":
                        inputs[k] = list_to_tree(v)
                elif isinstance(v, str):
                    v_clean = v.strip()
                    try:
                        if len(v_clean) == 32 and set(v_clean).issubset({'0', '1'}):
                            inputs[k] = int(v_clean, 2)
                        elif v_clean.isdigit() and k.lower() not in ["a", "b"]:
                            inputs[k] = int(v_clean)
                        else:
                            inputs[k] = v_clean
                    except:
                        inputs[k] = v_clean

            args = convert_arg_names(inputs, func)
            success, result_or_error = run_with_timeout(func, args=(), kwargs=args, timeout=2)

            if not success:
                raise result_or_error

            actual = result_or_error
            if actual is None:
                for name in ["nums", "matrix"]:
                    if name in args:
                        actual = args[name]
                        break

            expected = clean_strings(case["output"])

            # Special case: both result and in-place modification must match
            if isinstance(expected, str) and ',' in expected and any(x in expected for x in ['nums', 'matrix']):
                expected_num = int(expected.split(',')[0])
                expected_list = re.findall(r'\[(.*?)\]', expected)
                expected_list = [int(x) for x in expected_list[0].split(',') if x.strip().isdigit()] if expected_list else []
                result_ok = (actual == expected_num)
                actual_list = args.get("nums") or args.get("matrix")
                list_ok = (actual_list[:len(expected_list)] == expected_list)
                if result_ok and list_ok:
                    passed += 1
                    continue

            if smart_equal(actual, expected):
                passed += 1
            else:
                mismatches.append({
                    "input": case["input"],
                    "expected": case["output"],
                    "actual": actual
                })
        except Exception as e:
            mismatches.append({
                "input": case["input"],
                "expected": case["output"],
                "actual": f"Error: {e}"
            })

    return passed, len(test_cases), None, mismatches
````

````python
columns_to_evaluate = [
    "python_code",  # ← original reference Python code
    "parsed_python_from_javascript",
    "parsed_python_from_java",
    "parsed_python_from_cpp"
]

# Initialize final result table with IDs
final_df = df_filtered[["id"]].copy()

for col in columns_to_evaluate:
    print(f"Evaluating column: {col}")
    col_results = []

    for _, row in df_filtered.iterrows():
        qid = row["id"]
        raw_code = row[col]

        if not isinstance(raw_code, str) or not raw_code.strip():
            col_results.append((qid, 0, 0, 0.0, "No code", []))
            continue

        is_method = "class Solution" in raw_code
        parsed_code = unwrap_solution_class(raw_code) if is_method else raw_code

        json_path = os.path.join(json_folder, f"id{qid}.json")
        if not os.path.exists(json_path):
            col_results.append((qid, 0, 0, 0.0, "JSON not found", []))
            continue

        with open(json_path, "r") as f:
            test_data = json.load(f)
            func_name = get_function_name_from_code(parsed_code)
            if not func_name:
                col_results.append((qid, 0, len(test_data["test_cases"]), 0.0, "Function name not found", []))
                continue

        passed, total, error, mismatches = evaluate_function_robust(
            parsed_code, func_name, test_data["test_cases"], is_method
        )
        accuracy = passed / total if total > 0 else 0.0
        col_results.append((qid, passed, total, accuracy, error, mismatches))

    # Extract results
    qid_list, passed_list, total_list, accuracy_list, error_list, mismatch_list = zip(*col_results)

    if col == "python_code":
        prefix = "original"
    else:
        prefix = col.replace("parsed_python_from_", "")

    final_df[f"{prefix}_passed"] = passed_list
    final_df[f"{prefix}_total"] = total_list
    final_df[f"{prefix}_accuracy"] = accuracy_list
    final_df[f"{prefix}_error"] = error_list
    final_df[f"{prefix}_mismatches"] = mismatch_list
````
